```python

!pip install pandas numpy scikit-learn matplotlib seaborn tensorflow

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_squared_log_error
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense


df = pd.read_csv(io.BytesIO(uploaded['co2new.xlsx .csv']))



df = df.set_index('year').transpose()


def calculate_errors(true, pred):
    errors = {
        'RMSE': np.sqrt(mean_squared_error(true, pred)),
        'MAE': mean_absolute_error(true, pred),
        'MAPE': np.mean(np.abs((true - pred) / true)) * 100,
        'MFE': np.mean(pred - true),

    }
    return errors


results = {}
forecast_df = pd.DataFrame(index=range(2024, 2031))

for country in df.index:
    data = df.loc[country].dropna()

    if len(data) < 50:  # skip if too little data
        continue

    years = data.index.astype(int)
    emissions = data.values.reshape(-1, 1)


    train_years = years[years <= 2015]
    test_years = years[(years > 2015) & (years <= 2023)]

    train_data = emissions[years <= 2015]
    test_data = emissions[(years > 2015) & (years <= 2023)]


    scaler = MinMaxScaler()
    scaled_train = scaler.fit_transform(train_data)


    X_train = train_years.to_numpy().reshape(-1, 1)


    y_train = scaled_train


    model = Sequential()
    model.add(Dense(64, input_dim=1, activation='relu'))
    model.add(Dense(64, activation='relu'))
    model.add(Dense(1, activation='linear'))
    model.compile(optimizer='adam', loss='mse')
    model.fit(X_train, y_train, epochs=300, verbose=0)


    X_test = test_years.to_numpy().reshape(-1, 1)
    y_test_true = test_data
    y_test_pred_scaled = model.predict(X_test)
    y_test_pred = scaler.inverse_transform(y_test_pred_scaled)


    errors = calculate_errors(y_test_true, y_test_pred)
    results[country] = errors


    plt.figure(figsize=(12, 6))
    plt.plot(train_years, train_data, label='Train',color='black',linewidth=2)
    plt.plot(test_years, y_test_true, label='Actual',color='darkblue',linestyle='--')
    plt.plot(test_years, y_test_pred, label='Predicted',color='darkgreen',linestyle='--')
    plt.title(f"{country} - CO₂ Emission Prediction (ANN)")
    plt.xlabel("Year")
    plt.ylabel("CO₂ Emissions")
    plt.legend()
    plt.grid(True, linestyle='--', linewidth=0.8, alpha=0.5, color='gray')
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    plt.show()


    future_years = np.arange(2024, 2031).reshape(-1, 1)
    future_pred_scaled = model.predict(future_years)
    future_pred = scaler.inverse_transform(future_pred_scaled)
    forecast_df[country] = future_pred.flatten()

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)

results_df = pd.DataFrame(results).T
print("Evaluation Metrics (2016–2023):")
display(results_df)


forecast_df.index = np.arange(2024, 2031)
print("Forecasted Emissions (2024–2030):")
display(forecast_df)
```

LSTM

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_squared_log_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input
import io  # required for uploaded file handling in Colab

# Load and preprocess data
df = pd.read_csv(io.BytesIO(uploaded['co2new.xlsx .csv']))
df = df.set_index('year').transpose()

# Error metrics
def calculate_errors(true, pred):
    return {
        'RMSE': np.sqrt(mean_squared_error(true, pred)),
        'MAE': mean_absolute_error(true, pred),
        'MAPE': np.mean(np.abs((true - pred) / true)) * 100,
        'MFE': np.mean(pred - true),
        'MSLE': mean_squared_log_error(true, pred)
    }

# Sequence creation for LSTM
def create_sequences(data, window_size):
    X, y = [], []
    for i in range(len(data) - window_size):
        X.append(data[i:i + window_size])
        y.append(data[i + window_size])
    return np.array(X), np.array(y)

results = {}
forecast_df = pd.DataFrame(index=np.arange(2024, 2031))
window_size = 5  # Sequence window for LSTM

for country in df.index:
    series = df.loc[country].dropna()

    if len(series) < 50:
        continue

    years = series.index.astype(int)
    values = series.values.reshape(-1, 1)

    train_idx = years <= 2015
    test_idx = (years > 2015) & (years <= 2023)

    train_data = values[train_idx]
    test_data = values[test_idx]

    # Scaling
    scaler = MinMaxScaler()
    scaled_train = scaler.fit_transform(train_data)

    X_train, y_train = create_sequences(scaled_train, window_size)
    X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))

    # Build LSTM model
    model = Sequential()
    model.add(Input(shape=(window_size, 1)))
    model.add(LSTM(64, return_sequences=True))
    model.add(Dropout(0.2))
    model.add(LSTM(64))
    model.add(Dropout(0.2))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')

    # Train
    model.fit(X_train, y_train, epochs=200, verbose=0)

    # Prepare for prediction
    full_data = np.concatenate([train_data, test_data])
    scaled_full = scaler.transform(full_data)

    # Predict test (2016–2023)
    predictions = []
    input_seq = scaled_full[len(scaled_train) - window_size:len(scaled_train)].reshape(1, window_size, 1)

    for _ in range(len(test_data)):
        pred = model.predict(input_seq, verbose=0)[0][0]
        predictions.append(pred)
        pred = np.array(pred).reshape(1, 1, 1)
        input_seq = np.append(input_seq[:, 1:, :], pred, axis=1)

    pred_test = scaler.inverse_transform(np.array(predictions).reshape(-1, 1))
    errors = calculate_errors(test_data, pred_test)
    results[country] = errors

    # Plot results
    plt.figure(figsize=(10, 5))
    plt.plot(years[train_idx], train_data, label='Train',color='black',linewidth=2)
    plt.plot(years[test_idx], test_data, label='Actual',color='darkblue',linestyle='--')
    plt.plot(years[test_idx], pred_test, label='Predicted',color='darkgreen',linestyle='--')
    plt.title(f"{country} - CO₂ Emission Prediction (LSTM)")
    plt.xlabel("Year")
    plt.ylabel("CO₂ Emissions")
    plt.legend()
    plt.grid(True, linestyle='--', linewidth=0.8, alpha=0.5, color='gray')
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    plt.show()

    # Forecast future (2024–2030)
    future_preds = []
    input_seq = scaled_full[-window_size:].reshape(1, window_size, 1)

    for _ in range(7):  # for 2024–2030
        pred = model.predict(input_seq, verbose=0)[0][0]
        future_preds.append(pred)
        pred = np.array(pred).reshape(1, 1, 1)
        input_seq = np.append(input_seq[:, 1:, :], pred, axis=1)

    future_forecast = scaler.inverse_transform(np.array(future_preds).reshape(-1, 1))
    forecast_df[country] = future_forecast.flatten()


results_df = pd.DataFrame(results).T
print("Evaluation Metrics (2016–2023):")
display(results_df)


forecast_df.index = np.arange(2024, 2031)
print("Forecasted Emissions (2024–2030):")
display(forecast_df)
```

```python

!pip install pandas numpy scikit-learn matplotlib seaborn tensorflow openpyxl

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_squared_log_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, Dense, Dropout, Flatten, Input



df = pd.read_csv(io.BytesIO(uploaded['co2new.xlsx .csv']))
df = df.set_index('year').transpose()


def calculate_errors(true, pred):
    errors = {
        'RMSE': np.sqrt(mean_squared_error(true, pred)),
        'MAE': mean_absolute_error(true, pred),
        'MAPE': np.mean(np.abs((true - pred) / true)) * 100,
        'MFE': np.mean(pred - true),
        'MSLE': mean_squared_log_error(true, pred)
    }
    return errors


def create_sequences(data, window_size):
    X, y = [], []
    for i in range(len(data) - window_size):
        X.append(data[i:i + window_size])
        y.append(data[i + window_size])
    return np.array(X), np.array(y)


results = {}
forecast_df = pd.DataFrame(index=range(2024, 2031))
window_size = 5

for country in df.index:
    series = df.loc[country].dropna()

    if len(series) < 50:
        continue

    years = series.index.astype(int)
    values = series.values.reshape(-1, 1)


    train_idx = years <= 2015
    test_idx = (years > 2015) & (years <= 2023)

    train_data = values[train_idx]
    test_data = values[test_idx]


    scaler = MinMaxScaler()
    scaled_train = scaler.fit_transform(train_data)


    X_train, y_train = create_sequences(scaled_train, window_size)
    X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))


    model = Sequential()
    model.add(Input(shape=(window_size, 1)))
    model.add(Conv1D(64, kernel_size=2, activation='relu'))
    model.add(Dropout(0.2))
    model.add(Flatten())
    model.add(Dense(64, activation='relu'))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')
    model.fit(X_train, y_train, epochs=200, verbose=0)


    full_data = np.concatenate([train_data, test_data])
    scaled_full = scaler.transform(full_data)

    predictions = []
    input_seq = scaled_full[len(scaled_train)-window_size:len(scaled_train)].reshape(1, window_size, 1)

    for _ in range(len(test_data)):
        pred = model.predict(input_seq, verbose=0)[0]
        predictions.append(pred)
        new_val = np.array(pred).reshape(1, 1, 1)
        input_seq = np.concatenate([input_seq[:, 1:, :], new_val], axis=1)

    pred_test = scaler.inverse_transform(np.array(predictions).reshape(-1, 1))
    errors = calculate_errors(test_data, pred_test)
    results[country] = errors


    plt.figure(figsize=(10, 5))
    plt.plot(years[train_idx], train_data, label='Train',color='black',linewidth=2)
    plt.plot(years[test_idx], test_data, label='Actual',color='darkblue',linestyle='--')
    plt.plot(years[test_idx], pred_test, label='Predicted',color='darkgreen',linestyle='--')
    plt.title(f"{country} - CO₂ Emission Prediction (CNN)")
    plt.xlabel("Year")
    plt.ylabel("CO₂ Emissions")
    plt.legend()
    plt.grid(True, linestyle='--', linewidth=0.8, alpha=0.5, color='gray')
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    plt.show()


    future_preds = []
    input_seq = scaled_full[-window_size:].reshape(1, window_size, 1)

    for _ in range(7):
        pred = model.predict(input_seq, verbose=0)[0]
        future_preds.append(pred)
        new_val = np.array(pred).reshape(1, 1, 1)
        input_seq = np.concatenate([input_seq[:, 1:, :], new_val], axis=1)

    future_forecast = scaler.inverse_transform(np.array(future_preds).reshape(-1, 1))
    forecast_df[country] = future_forecast.flatten()

# Final Results
results_df = pd.DataFrame(results).T
print("Evaluation Metrics (2016–2023):")
display(results_df)

forecast_df.index = np.arange(2024, 2031)
print("Forecasted Emissions (2024–2030):")
display(forecast_df)
```
